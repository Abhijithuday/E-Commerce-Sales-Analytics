Data preprocessing is performed to make the dataset suitable for analysis.

<h1 style="text-align: left;font-size: 18px;font-weight: bold;">
    Convert Date Column
</h1>

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
print(df['order_date'].dtype)
df['order_date'].head()
```

<h1 style="text-align: left;font-size: 18px;font-weight: bold;">
    Check Date Conversion
</h1>

```python
invalid_dates = df['order_date'].isnull().sum()

print("Invalid dates:", invalid_dates)
```

<h1 style="text-align: left;font-size: 18px;font-weight: bold;">
    Remove Duplicates
</h1>

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

<h1 style="text-align: left;font-size: 18px;font-weight: bold;">
    Handle Missing Values
</h1>

```python
numerical_columns = df.select_dtypes(include=np.number).columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print(df.isnull().sum())
```

## Derived Columns

Additional columns will be created to improve the analysis.

New columns include:

- Gross Sales
- Discount Amount
- Net Revenue
- Revenue Per Unit
- Year
- Month
- Month Name
- Day of Week
- Delivery Category
- Rating Category

```python
# Gross sales before discount

df['gross_sales'] = df['quantity'] * df['unit_price']

# Discount amount

df['discount_amount'] = df['gross_sales'] * df['discount']

# Revenue per unit

df['revenue_per_unit'] = df['revenue'] / df['quantity']

# Extract date information

df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df['day_of_week'] = df['order_date'].dt.day_name()

# Delivery category

df['delivery_category'] = pd.cut(
    df['delivery_days'],
    bins=[0, 3, 7, 100],
    labels=['Fast', 'Normal', 'Slow']
)

# Rating category

df['rating_category'] = pd.cut(
    df['customer_rating'],
    bins=[0, 2, 3, 4, 5],
    labels=['Poor', 'Average', 'Good', 'Excellent']
)

df.head()
```

```python
# Calculate major KPIs

total_orders = df['order_id'].nunique()
total_customers = df['customer_id'].nunique()
total_revenue = df['revenue'].sum()
average_order_value = df['revenue'].mean()
total_quantity = df['quantity'].sum()
average_rating = df['customer_rating'].mean()
average_delivery = df['delivery_days'].mean()
average_discount = df['discount'].mean()

print("========== E-COMMERCE SALES KPIs ==========")

print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Revenue:", round(total_revenue, 2))
print("Average Order Value:", round(average_order_value, 2))
print("Total Quantity Sold:", total_quantity)
print("Average Customer Rating:", round(average_rating, 2))
print("Average Delivery Days:", round(average_delivery, 2))
print("Average Discount:", round(average_discount * 100, 2), "%")
```

    Automated Insight Generation

```python
# Category with highest revenue

category_revenue = df.groupby('product_category')['revenue'].sum()

best_category = category_revenue.idxmax()

# Region with highest revenue

region_revenue = df.groupby('region')['revenue'].sum()

best_region = region_revenue.idxmax()

# Most popular payment method

best_payment = df['payment_method'].value_counts().idxmax()

# Highest rated category

category_rating = df.groupby('product_category')['customer_rating'].mean()

highest_rated_category = category_rating.idxmax()

# Fastest average delivery region

region_delivery = df.groupby('region')['delivery_days'].mean()

fastest_region = region_delivery.idxmin()

print("========== KEY FINDINGS ==========")

print("1. Highest revenue product category:", best_category)

print("2. Highest revenue region:", best_region)

print("3. Most commonly used payment method:", best_payment)

print("4. Highest average rated category:", highest_rated_category)

print("5. Region with lowest average delivery time:", fastest_region)
```
